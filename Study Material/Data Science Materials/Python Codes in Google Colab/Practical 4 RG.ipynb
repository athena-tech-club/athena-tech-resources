```python
#Again, bruteforce but this time, we are going to scan the start city
#Read the data
import pandas as pd
import time
mat=pd.read_csv("/content/11citytsp.csv", header=None).values
#Checking
print(mat)
from itertools import permutations
#asking for the starting city
startcity=int(input("Enter the start city from 0 to " + str(mat.shape[0]-1) + ":"))
#generating city names
citynames=list(range(mat.shape[0]))#rowsXcolumns,shape[0]=rows
#removing the starting city
citynames.remove(startcity) #startcity is scanned, so, we need to generate permutat
#generating permutations
per=list(permutations(citynames))
#print(per)
st=time.process_time()
besttourlength=np.inf
for tour in per:
  tourlength=0
  for i in range(len(tour)-1):
    tourlength+=mat[tour[i],tour[i+1]]
    tourlength+=mat[tour[i+1],startcity] #lastcity to startcity
    tourlength+=mat[startcity,tour[0]] #startcity to actual second city
    if tourlength < besttourlength: #minimum logic
      besttourlength=tourlength
      besttour=tour
      besttour=list(besttour)
      besttour.insert(0,startcity) #inserting startcity
et=time.process_time()
time_taken_ms=(et-st)*1000
print("Best Tour:", besttour)
print("Best Tour's Length:", besttourlength)
print("Time Taken:",time_taken_ms)
```

```python
#TSP using Nearest Neighbour Heuristic, not always optimal
#Read the data
import pandas as pd
import time
mat=pd.read_csv("/content/11citytsp.csv", header=None).values.astype(float)
#asking for the starting city
startcity=int(input("Enter the start city from 0 to " + str(mat.shape[0]-1) + ":"))
st=time.process_time()
tourlength=0
besttour=[startcity] #startcity is already scanned
mat[mat==0]=np.inf #cities x-x are marked as Inf
print(mat)
matorigcopy=mat.copy() #at the end, we will be required to come back to startcity
for i in range(mat.shape[0]-1):
    if i==0:
      tourlength+=min(mat[startcity,:])
      nextbestind=np.argmin(mat[startcity,:])
      besttour.append(nextbestind)
      mat[:,nextbestind]=np.inf #can't visit again
      mat[:,startcity]=np.inf #can't visit again except at the end, we will be re
    else:
      tourlength+=min(mat[nextbestind,:])
      nextbestind=np.argmin(mat[nextbestind,:])
      mat[:,nextbestind]=np.inf #can't visit again
      besttour.append(nextbestind)

tourlength+=matorigcopy[nextbestind,startcity] #lastcity to startcity
et=time.process_time()
time_taken_ms=(et-st)*1000
print("Best Tour:", besttour)
print("Best Tour's Length:", tourlength)
print("Time Taken:",time_taken_ms)
```

```python
import itertools

# Function to calculate the total distance for a given path
def total_distance(path, distances):
    total = 0
    for i in range(len(path) - 1):
        total += distances[path[i]][path[i + 1]]
    total += distances[path[-1]][path[0]]
     # Add the distance from the last city back to the starting city
    return total

# Function to find the shortest path using brute force
def brute_force_tsp(cities, distances):
    shortest_path = None
    shortest_distance = float('inf')
    for perm in itertools.permutations(cities):
        dist = total_distance(perm, distances)
        if dist < shortest_distance:
            shortest_distance = dist
            shortest_path = perm
    return shortest_path, shortest_distance

# Define the cities and their distances
cities = ['p', 'q', 'r', 's']
distances = {
    'p': {'p': 0, 'q': 3, 'r': 6, 's': 8},
    'q': {'p': 3, 'q': 0, 'r': 9, 's': 4},
    'r': {'p': 6, 'q': 9, 'r': 0, 's': 2},
    's': {'p': 8, 'q': 4, 'r': 2, 's': 0}
}

# Solve the TSP using brute force
shortest_path, shortest_distance = brute_force_tsp(cities, distances)

# Print the result
print("Shortest Path:", shortest_path)
print("Shortest Distance:", shortest_distance)
```
