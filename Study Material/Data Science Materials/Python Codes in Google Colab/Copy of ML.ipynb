```python
#Multiple Linear Regression using Normal Equation Method
import numpy as np
from sklearn import datasets, metrics
from numpy.linalg import inv, pinv, LinAlgError

X, y = datasets.load_boston(return_X_y=True)

X_train_temp1=X[0:400,:]
X_train=np.ones((X_train_temp1.shape[0],X_train_temp1.shape[1]+1))
X_train[:,1:]=X_train_temp1
print("Type of X_train:", type(X_train), "Shape of X_train:", X_train.shape)

y_train=y[0:400]

X_test_temp1=X[400:506,:]
X_test=np.zeros((X_test_temp1.shape[0],X_test_temp1.shape[1]+1))
X_test[:,0]=np.ones((X_test_temp1.shape[0]))
X_test[:,1:]=X_test_temp1
print("Type of X_test:", type(X_test), "Shape of X_test:", X_test.shape)


y_test=y[400:506]
```

Equation:



```python
theta=np.zeros(X_train.shape[1])

try:
    XTXi=inv(np.dot(X_train.T,X_train))
except LinAlgError:
    XTXi=pinv(np.dot(X_train.T,X_train))

XTy=np.dot(X_train.T,y_train)

theta=np.dot(XTXi,XTy)

print("Thetas:", theta)
print("Thetas Shape:", theta.shape)

predictions=np.dot(theta,X_test.T)

print("MAE:", metrics.mean_absolute_error(y_true=y_test,y_pred=predictions))
print("MSE:", metrics.mean_squared_error(y_true=y_test,y_pred=predictions))
```

```python
#Multiple Linear Regression using Gradient Descent
import numpy as np
from sklearn import datasets, metrics
from sklearn.preprocessing import StandardScaler

X, y = datasets.load_boston(return_X_y=True)

X_train_temp1=X[0:400,:]
X_train=np.ones((X_train_temp1.shape[0],X_train_temp1.shape[1]+1))
X_train[:,1:]=X_train_temp1
print("Type of X_train:", type(X_train), "Shape of X_train:", X_train.shape)

y_train=y[0:400]


X_test_temp1=X[400:506,:]
X_test=np.zeros((X_test_temp1.shape[0],X_test_temp1.shape[1]+1))
X_test[:,0]=np.ones((X_test_temp1.shape[0]))
X_test[:,1:]=X_test_temp1
print("Type of X_test:", type(X_test), "Shape of X_test:", X_test.shape)


y_test=y[400:506]


scaler=StandardScaler()
scaler.fit(X_train[:,1:])
X_train[:,1:]=scaler.transform(X_train[:,1:])
X_test[:,1:]=scaler.transform(X_test[:,1:])
```

```python
theta=np.random.uniform(0,1,size=(X_train.shape[1]))
print("Type of theta:", type(theta), "Shape of Theta:", theta.shape)


niterations=1000
alpha=0.01
m=X_train.shape[0]
n=X_train.shape[1]

for i in range(niterations):
    update=np.zeros(X_train.shape[1])
    ypred=np.dot(X_train,theta)
    error=ypred - y_train
    for j in range(n):
        update[j]=np.sum(error*(X_train.T)[j])
    theta = theta - (1/m)*(alpha)*update


print("Theta:",theta)
print("Theta Shape:", theta.shape)

predictions=np.dot(X_test, theta)

-print("MAE:", metrics.mean_absolute_error(y_true=y_test,y_pred=predictions))
print("MSE:", metrics.mean_squared_error(y_true=y_test,y_pred=predictions))
```

```python
#Implementation of Incremental Mathematical Stream Regression
from sklearn import metrics
import pandas as pd

count=0

testdataframe=pd.read_csv("boston.csv",header=None)
testdataarray=testdataframe.values
X_test=testdataarray[400:506,:-1]
y_test=testdataarray[400:506,-1]

def myGeneratorfun(wsize=100):
    print("In myGeneratorfun:")
    counter=0
    for chunk in pd.read_csv('boston.csv', header=None, chunksize=wsize):
        print("In for loop, counter vlaue is", counter)
        chunk_array=chunk.values
        counter=counter+1
        yield (chunk_array[:,:-1], chunk_array[:,-1])
        if counter >= 4:
            break
```

Regression Parameters:


Regression Parameters for the kth window:




where M is sum of X'X products and V is sum of X'y products

```python
myGenerator = myGeneratorfun(100)

M=np.zeros((X_test.shape[1],X_test.shape[1]))
V=np.zeros((X_test.shape[1]))
Beta=np.zeros(X_test.shape[1])

for (x,y) in myGenerator:

    try:
        XTXi=inv(M + np.dot(x.T,x))
    except LinAlgError:
        XTXi=pinv(M + np.dot(x.T,x))



    XTy=np.dot(x.T,y)


    Beta=np.dot(XTXi,(V + XTy))
    M=M+np.dot(x.T,x)
    V=V+XTy

    #print("Betas:", Beta)
    #print("Betas Shape:", Beta.shape)

    predictions=np.dot(Beta,X_test.T)
    print(predictions.shape)
    print("MAE:", metrics.mean_absolute_error(y_true=y_test,y_pred=predictions))
    print("MSE:", metrics.mean_squared_error(y_true=y_test,y_pred=predictions))
```
