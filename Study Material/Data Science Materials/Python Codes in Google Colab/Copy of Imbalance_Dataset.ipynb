```python
cancer_dataset["Status"] = cancer_dataset["Status"].replace({"Alive":0,"Dead":1})
cancer_dataset["Progesterone Status"] = cancer_dataset["Progesterone Status"].replace({"Positive":0,"Negative":1})
cancer_dataset["Estrogen Status"] = cancer_dataset["Estrogen Status"].replace({"Positive":0,"Negative":1})
cancer_dataset["A Stage"] = cancer_dataset["A Stage"].replace({"Regional":0,"Distant":1})
cancer_dataset["Grade"] = cancer_dataset["Grade"].replace({"Well differentiated; Grade I":0,"Moderately differentiated; Grade II":1,"Poorly differentiated; Grade III":2,"Undifferentiated; anaplastic; Grade IV":3})
#cancer_dataset["Race"] = cancer_dataset["Race"].replace({"White":0,"Black":1,"Other (American Indian/AK Native, Asian/Pacific Islander)":2})
cancer_dataset["Marital Status"] = cancer_dataset["Marital Status"].replace({'Married (including common law)':0,'Divorced':1,'Single (never married)':2,
'Widowed':3,'Separated':4})
#cancer_dataset["T Stage"] = cancer_dataset["T Stage"].replace({'T2':1,'T1':0,'T3':2,'T4':3})
cancer_dataset["N Stage"] = cancer_dataset["N Stage"].replace({'N3':2,'N2':1,'N1':0})
cancer_dataset["6th Stage"] = cancer_dataset["6th Stage"].replace({'IIIC':4,'IIIA':2,'IIB':1,'IIA':0,'IIIB':3})
print(cancer_dataset.head())
```

## Data splitting

```python
from sklearn.model_selection import train_test_split

X = cancer_dataset.drop("Status",axis=1)
y = cancer_dataset["Status"]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3)
```

## Data Standardizing

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.fit_transform(X_test)
print(X_train)
```

## Model building

```python
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
```

### Without balancing data

Linear SVM

```python
svc = SVC() ###rbf is default kernel
clf = svc.fit(X_train,y_train)
y_pred = svc.predict(X_test)
print("Confusion Matrix: ")
print(confusion_matrix(y_test, y_pred))
print("Classifiction Report :")
print(classification_report(y_test, y_pred))
```

Logistic Reg

```python
lr = LogisticRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
print("Confusion Matrix: ")
print(confusion_matrix(y_test, y_pred))
print("Classifiction Report :")
print(classification_report(y_test, y_pred))
```

Naiva bayes - GaussianNb

```python
gb = GaussianNB()
gb.fit(X_train,y_train)
y_pred = gb.predict(X_test)
print("Confusion Matrix: ")
print(confusion_matrix(y_test, y_pred))
print("Classifiction Report :")
print(classification_report(y_test, y_pred))
```

DecisionTreeClassifier

```python
dtree = DecisionTreeClassifier()
dtree.fit(X_train,y_train)
y_pred = dtree.predict(X_test)
print("Confusion Matrix: ")
print(confusion_matrix(y_test, y_pred))
print("Classifiction Report :")
print(classification_report(y_test, y_pred))
```
