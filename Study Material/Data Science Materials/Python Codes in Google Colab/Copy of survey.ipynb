```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
data1=pd.read_csv("surveystudent.csv",header ='infer').values
data2=pd.read_csv("surveyfaculty.csv", header = 'infer').values
X1=data1[:,2:]
X2=data2[:,2:]
test_split = float(input('Enter the test size: '))
X1_train,X1_test=train_test_split(X1,test_size=test_split)
X2_train,X2_test=train_test_split(X2, test_size = test_split)
k = int(input('Enter the number of clusters you want to divide in: '))
n = int(input('Enter number of iterations you want to run: '))
centroids1 = np.zeros(shape=(k,X1_train.shape[1]))
centroids2 = np.zeros(shape = (k, X2_train.shape[1]))
per1 = np.random.permutation(X1_train.shape[0])
per2 = np.random.permutation(X2_train.shape[0])
for i in range(k):
    centroids1[i,:]=X1_train[per1[i],:]
    centroids2[i,:]=X2_train[per2[i],:]
print(centroids1)
print(centroids2)
for i in range(n):
    dist1 = np.zeros(shape=(k,X1_train.shape[0]))
    dist2 = np.zeros(shape=(k,X2_train.shape[0]))
    for i in range(k):
        dist1[i,:]=np.sqrt(np.sum((X1_train-centroids1[i])**2,axis=1).astype(int))
        dist2[i,:]=np.sqrt(np.sum((X2_train-centroids2[i])**2, axis = 1).astype(int))
    membership1 = np.argmin(dist1,axis=0)
    membership2 = np.argmin(dist2,axis=0)
    for i in range(k):
        centroids1[i,:]=np.mean(X1_train[membership1==i,:],axis=0)
        centroids2[i,:]=np.mean(X2_train[membership2==i,:],axis=0)
print(centroids1)
print(centroids2)
import matplotlib.pyplot as plt
print("Strongly Disagree")
X1 = centroids1[0]
X2 = centroids2[0]
plt.plot(X1,label = "Student", marker = 'o')
plt.plot(X2,label = "Faculty", marker = 's')
plt.legend()
plt.show()
print("Disagree")
X1 = centroids1[1]
X2 = centroids2[1]
plt.plot(X1,label = "Student", marker = 'o')
plt.plot(X2,label = "Faculty", marker = 's')
plt.legend()
plt.show()
print("Neutral")
X1 = centroids1[2]
X2 = centroids2[2]
plt.plot(X1,label = "Student", marker = 'o')
plt.plot(X2,label = "Faculty", marker = 's')
plt.legend()
plt.show()
print("Agreed")
X1 = centroids1[3]
X2 = centroids2[3]
plt.plot(X1,label = "Student", marker = 'o')
plt.plot(X2,label = "Faculty", marker = 's')
plt.legend()
plt.show()
print("Strongly Agreed")
X1 = centroids1[4]
X2 = centroids2[4]
plt.plot(X1,label = "Student", marker = 'o')
plt.plot(X2,label = "Faculty", marker = 's')
plt.legend()
plt.show()
```
