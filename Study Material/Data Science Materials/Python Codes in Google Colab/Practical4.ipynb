```python
import pandas as pd
import time
import numpy as np
mat= pd.read_csv("6citytsp.csv", header=None).values.astype(float)
startcity = int(input("Enter the start city for 0 to " + str(mat.shape[0]-1) + ": "))
st = time.process_time()
tourlength = 0
besttour =[startcity]
mat[mat==0] = np.inf
matorigycopy = mat.copy()
print(mat)
for i in range(0,mat.shape[0]-1):
    if i==0:
        tourlength += np.min(mat[startcity, :])
        nextbestind = np.argmin(mat[startcity,:])
        besttour.append(nextbestind)
        mat[:,nextbestind] = np.inf
        mat[:,startcity] = np.inf
    else:
        tourlength += np.min(mat[nextbestind,:])
        nextbestind = np.argmin(mat[nextbestind,:])
        besttour.append(nextbestind)
        mat[:,nextbestind] = np.inf
tourlength += matorigycopy[nextbestind,startcity]
et = time.process_time()
time_taken_ms = (et-st)*1000
print("Time taken : ", time_taken_ms)
print("Best Tour: ", besttour)
print("Best Tour length : ", tourlength)
```

```python
import pandas as pd
import time
from itertools import permutations
mat = pd.read_csv("6citytsp.csv", header = None).values

citynames = list(range(mat.shape[0]))
per = permutations(citynames)
print(type(per))
per = list(per)
print(type(per))
print(per)
import numpy as np
st = time.process_time()
besttourlength = np.inf
for tour in per:
    tourlength = 0
for i in range(len(tour)-1):
    tourlength += mat[tour[i],tour[i+1]]
    tourlength += mat[tour[i+1],tour[0]]
    if tourlength<besttourlength:
        besttourlength = tourlength
        besttour = tour
        besttour = list(besttour)
et = time.process_time()
timetaken_ms = (et-st)*1000
print("best tour length: ",besttourlength)
print("Best tour : ",besttour)
print("Time Taken : ",timetaken_ms)
```
