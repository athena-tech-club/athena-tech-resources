```python

n = int(input("Enter how many number you want to work with : "))
import numpy as np
my_list = []
for i in range(n) :
  num = float(input("Enter element " + str(i+1) + " : "))
  my_list.append(num)
print(my_list)
list1 = np.array(my_list)
print(list1)

#for mean with library
print("Finding Mean using Library : \n")
mean = np.mean(list1)
print("Mean = ", mean)

#for mean without library
print("Finding Mean Without using Library : \n")
sum = 0
for i in list1 :
  sum+=i
  mean = sum/n
print("Mean = ", mean)

#for median with library
print("Finding Median with using Library : \n")
med = np.median(list1)
print("Median = ", med)

#for median without library
print("Finding Median without using Library : \n")
sorted_list = np.sort(list1)
if (n%2 == 0):
  print("Median = ", ((sorted_list[n//2] +
  sorted_list[(n//2)+1])/2))
else :
  print("Median = ", (sorted_list[n//2]))

#for variance with library
print("Finding variance with using Library : \n")
var = np.var(list1)
print("Variance = ", var)

#for variance without library
print("Finding variance without using Library : \n")
s = 0
for i in sorted_list :
  s += ((i-mean)**2)
print("Variance = ", (s/n))

#for standard variance with library
print("Finding Standard Variance with using Library : \n")
std = np.std(list1)
print("Standard Variance = ", std)

#for standard variance without library
print("Finding Standard Variance without using Library :\n")
std = var**(1/2)
print("Standard Variance = ", std)

#for mode with library
print("Finding Mode with using Library : \n")
import statistics as st
mode = st.mode(sorted_list)
print("Mode = ", mode)

#for mode without library
print("Finding Mode without using Library : \n")
l1 = []
l2 = []
for i in sorted_list:
  if i not in l1:
    l1.append(i)
  else :
    l2.append(i)
print("Mode = ", set(l2))
```
