```python
from sklearn.model_selection import train_test_split
from sklearn import datasets,tree
from sklearn.metrics import confusion_matrix,accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
```

```python
X,y=datasets.load_iris(return_X_y=True)
X_Train,X_Test,y_Train,y_Test=train_test_split(X,y,test_size=0.2,random_state=40,stratify=y)
```

```python
id3=DecisionTreeClassifier(criterion='entropy',splitter='best')
id3.fit(X_Train,y_Train)
pred=id3.predict(X_Test)
print("accuracy_score is : ",accuracy_score(y_Test,pred))
print("Confusion matrix is: \n",confusion_matrix(y_Test,pred))
```

```python
CART = tree.DecisionTreeClassifier()
CART.fit(X, y)
pred=CART.predict(X_Test)
print("accuracy_score is : ",accuracy_score(y_Test,pred))
print("Confusion matrix is: \n",confusion_matrix(y_Test,pred))
```

```python

#Loading the libraries and importing them
import numpy as np
from sklearn import datasets,metrics
from sklearn.tree import DecisionTreeClassifier
#loads previously installed boston database
X , y = datasets.load_iris(return_X_y=True)
#Splitting the data into train_data, train_target, test_data, test_target
#for training set we took even no. of examples
X_train=X[range(1,150,2),:]
y_train=y[range(1,150,2)]
#for test set we took odd no. of examples
X_test=X[range(0,150,2),:]
y_test=y[range(0,150,2)]
#Applying the Decision Tree Classifier with selecting the criteria as entropy
#Default value is gini
clf=DecisionTreeClassifier(criterion='entropy')
#Fitting the train and test data
clf.fit(X_train,y_train)
#predict using training dataset
prediction = clf.predict(X_test)
print("prediction : ",prediction)
print("Accuracy = ", metrics.accuracy_score(y_test,prediction,normalize=True))
print(metrics.classification_report(y_test,prediction))
print(metrics.confusion_matrix(y_test,prediction))
```

```python
from sklearn import tree
text_representation = tree.export_text(clf)
print(text_representation)
```

```python
#Loading the libraries and importing them
import numpy as np
from sklearn import datasets,metrics
from sklearn.tree import DecisionTreeClassifier
#loads previously installed boston database
X , y = datasets.load_iris(return_X_y=True)
#for training set we took even no. of examples
X_t=X[range(0,150,2),:]
y_t=y[range(0,150,2)]
#for test set we took odd no. of examples
X_t1=X[range(1,150,2),:]
y_t1=y[range(1,150,2)]
#Applying the Decision Tree Classifier with selecting the criteria as entropy
#Default value is gini
clf=DecisionTreeClassifier(criterion='gini')
# clf=DecisionTreeClassifier(criterion='gini',max_depth=4)
#Fitting the train and test data
clf.fit(X_t,y_t)
#predict using training dataset
prediction = clf.predict(X_t1)
print("prediction : ",prediction)
print("Accuracy = ", metrics.accuracy_score(y_t1,prediction,normalize=True))
print(metrics.classification_report(y_t1,prediction))
print(metrics.confusion_matrix(y_t1,prediction))
```

With GRID SEARCH

```python
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix
tree_para = {'criterion':['gini','entropy'],'max_depth':[4,5,6,7,8,9,10,11,12,15,20,25]}
clf = GridSearchCV(DecisionTreeClassifier(), tree_para, cv=5,refit= True,verbose = 3)
clf.fit(X_t,y_t)
print(clf.best_params_)
# print how our model looks after hyper-parameter tuning  b
print(clf.best_estimator_)
grid_predictions = clf.predict(X_t1)
# print classification report
print(classification_report(y_t1, grid_predictions))
```
