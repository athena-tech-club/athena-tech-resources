```python
#K-means Clustering of IRIS Dataset

import numpy
from sklearn import datasets, metrics
from sklearn.cluster import KMeans

data, target=datasets.load_iris(return_X_y=True)

X_train=data[range(0,150,2),:]
X_test=data[range(1,150,2),:]

model=KMeans(n_clusters=3) #max_iter = 300

model.fit(X_train)

print(model.labels_)

prediction = model.predict(X_test)

print(prediction)

#prediction = numpy.choose(prediction, ['ir_se','ir_ve','ir_vr'])
#prediction = numpy.choose(prediction, [0,1,2]).astype(numpy.int64)

#print(prediction)



print(model.cluster_centers_)


#print(metrics.accuracy_score(target, prediction, normalize=True))

#print(metrics.confusion_matrix(target, prediction))
```

```python
#MLPClassifier For IRIS Flower Classification
import numpy
from sklearn import datasets, metrics
from sklearn.neural_network import MLPClassifier

X, y=datasets.load_iris(return_X_y=True)

X_train=X[range(0,150,2),:]   #iris.data is numpy array
y_train=y[range(0,150,2)]

X_test=X[range(1,150,2),:]
y_test=y[range(1,150,2)]

clf = MLPClassifier(solver='sgd', hidden_layer_sizes=(65), max_iter=200) #creating instance of the classifier

# activation : {'identity', 'logistic', 'tanh', 'relu'}, default 'relu'
# learning_rate : {'constant', 'invscaling', 'adaptive'}, default 'constant'
#learning_rate_init : double, optional, default 0.001
# solver{'lbfgs', 'sgd', 'adam'}, default='adam'

clf.fit(X_train,y_train)
#print(clf.get_params())
prediction = clf.predict(X_test) #predict using the learnt classifier

print(prediction)
print(clf.predict_proba(X_test));
print(metrics.accuracy_score(y_test, prediction, normalize=True))
print(metrics.classification_report(y_test, prediction))
print(metrics.confusion_matrix(y_test, prediction))
```

```python
import numpy as np
from sklearn import svm, datasets, metrics

X, y = datasets.load_iris(return_X_y=True)

X_train=X[range(0,150,2),:]
y_train=y[range(0,150,2)]

X_test=X[range(1,150,2),:]
y_test=y[range(1,150,2)]

clf = svm.LinearSVC() #creating instance of a classifier
# LinearSVC does not accept keyword kernel, as this is assumed to be linear.
# It also lacks some of the members of SVC and NuSVC
# dual: bool, (default=True)
# multi_class: str, ‘ovr’ or ‘crammer_singer’ (default=’ovr’)
# max_iter: int, (default=1000)


# or clf = svm.SVC() or clf = svm.NuSVC()
# regr = svm.LinearSVR() regr = svm.SVC() or regr = svm.NuSVR #()in case of regression
# SVC and NuSVC use 'ovo' (one vs one) strategy. It is the only option availalbe
# in SVC and NuSVC for multiclass classification.

clf.fit(X_train, y_train)

prediction = clf.predict(X_test) #predict using the learnt classifier

print(prediction)
print(metrics.accuracy_score(y_test, prediction, normalize=True))
print(metrics.classification_report(y_test, prediction))
print(metrics.confusion_matrix(y_test, prediction))
```

```python
import numpy
from sklearn import neighbors, datasets, metrics
from sklearn.model_selection import GridSearchCV

X, y=datasets.load_iris(return_X_y=True) #Loading the Iris Dataset

X_train=X[range(0,150,2),:]
y_train=y[range(0,150,2)]

X_test=X[range(1,150,2),:]
y_test=y[range(1,150,2)]

#creating instance of a classifier
clf = neighbors.KNeighborsClassifier()

#param_grid should be dictionary or list of dictionaries
#Specify hyperparameters which you want to tune

#If some hyperparameter you don't want to tune, specify that parameter
#while creating instance of the classifier

param_grid = {'n_neighbors': [1, 5, 10, 20], 'weights': ['uniform', 'distance']}

clf_gs=GridSearchCV(clf,param_grid,cv=5)
#there is a refit argument whose default is True

clf_gs.fit(X_train, y_train)

#Print best parameters
print(clf_gs.best_params_)

prediction = clf_gs.predict(X_test) #predict using the learnt classifier

print("############### Predictions #################")
print(prediction)
print("#############################################")

print("Accuracy = ", metrics.accuracy_score(y_test, prediction, normalize=True))

print(metrics.classification_report(y_test, prediction))

print(metrics.confusion_matrix(y_test, prediction))
```
