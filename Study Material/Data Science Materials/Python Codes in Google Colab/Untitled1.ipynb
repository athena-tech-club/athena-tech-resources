```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Generate some random data
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Initialize centroids
k = 4
centroids = X[np.random.choice(len(X), k, replace=False)]

# Function to assign clusters
def assign_clusters(X, centroids):
    distances = np.sqrt(((X - centroids[:, np.newaxis])**2).sum(axis=2))
    return np.argmin(distances, axis=0)

# Function to update centroids
def update_centroids(X, clusters, k):
    new_centroids = np.zeros((k, X.shape[1]))
    for i in range(k):
        new_centroids[i] = np.mean(X[clusters == i], axis=0)
    return new_centroids

# K-means clustering
max_iter = 100
for _ in range(max_iter):
    clusters = assign_clusters(X, centroids)
    new_centroids = update_centroids(X, clusters, k)
    if np.allclose(new_centroids, centroids):
        break
    centroids = new_centroids

# Plot the clusters
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='viridis', s=50, alpha=0.5, label='Visual')
plt.scatter(X[:, 0], X[:, 1], c=clusters, cmap='viridis', s=50, alpha=0.5, label='audio')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, label='Centroids')
plt.title('K-means Clustering')
plt.legend()
plt.show()

```

```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Generate some random data
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Select only the points with positive x-values
X_positive_x_axis = X[X[:, 0] > 0]

# Initialize centroids for the positive x-axis data
k = 4
centroids_positive_x_axis = X_positive_x_axis[np.random.choice(len(X_positive_x_axis), k, replace=False)]

# Function to assign clusters
def assign_clusters(X, centroids):
    distances = np.sqrt(((X - centroids[:, np.newaxis])**2).sum(axis=2))
    return np.argmin(distances, axis=0)

# Function to update centroids
def update_centroids(X, clusters, k):
    new_centroids = np.zeros((k, X.shape[1]))
    for i in range(k):
        new_centroids[i] = np.mean(X[clusters == i], axis=0)
    return new_centroids

# K-means clustering
max_iter = 100
for _ in range(max_iter):
    clusters_positive_x_axis = assign_clusters(X_positive_x_axis, centroids_positive_x_axis)
    new_centroids_positive_x_axis = update_centroids(X_positive_x_axis, clusters_positive_x_axis, k)
    if np.allclose(new_centroids_positive_x_axis, centroids_positive_x_axis):
        break
    centroids_positive_x_axis = new_centroids_positive_x_axis

# Plot the clusters on the positive x-axis
plt.scatter(X_positive_x_axis[:, 0], X_positive_x_axis[:, 1], c=clusters_positive_x_axis, cmap='viridis', s=50, alpha=0.5, label='Visual')
plt.scatter(X_positive_x_axis[:, 0], X_positive_x_axis[:, 1], c=clusters_positive_x_axis, cmap='viridis', s=50, alpha=0.5, label='')
plt.scatter(centroids_positive_x_axis[:, 0], centroids_positive_x_axis[:, 1], c='red', marker='X', s=200, label='Centroids')
plt.title('K-means Clustering on the Positive X-axis')
plt.legend()
plt.show()
```
