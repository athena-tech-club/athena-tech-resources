## **DEMONSTRATION OF TOKENIZER**

```python
# 1. Importing Libraries
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# 2. Loading the Dataset
# This could be replaced with actual SMS data (like the SMS Spam Collection dataset)

# Sample data for demonstration
data = {'Message': ['Free entry in 2 a wkly comp to win FA Cup final tkts 21st May 2023.',
                    'Nah I don’t think he goes to usf, he lives around here though',
                    'Win a cash prize of 1000. Click here to claim.',
                    'Even my brother is not like to speak with me.'],
        'Label': ['spam', 'ham', 'spam', 'ham']}

df = pd.DataFrame(data)

# 3. Data Preprocessing
# Convert labels to binary (spam: 1, ham: 0)
df['Label'] = df['Label'].map({'spam': 1, 'ham': 0})

# 4. Text Vectorization
# Convert text into a bag-of-words model
vectorizer = CountVectorizer() # converts documents into a matrix of token counts
#[[1, 1, 1, 1, 1, 0, 0],   # 'Free entry in a contest'
# [1, 0, 0, 1, 0, 1, 1]]   # 'Win a free ticket'

X = vectorizer.fit_transform(df['Message'])

# 5. Splitting the Dataset
# Splitting the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, df['Label'], test_size=0.3, random_state=42)

# 6. Model Training
# Initializing and training the Multinomial Naive Bayes model
model = MultinomialNB()
model.fit(X_train, y_train)

# 7. Making Predictions
# Predicting the test data
y_pred = model.predict(X_test)

# 8. Evaluating the Model
# Checking the accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy * 100:.2f}%')

# Displaying the confusion matrix
print('Confusion Matrix:')
print(confusion_matrix(y_test, y_pred))

# Classification report
print('Classification Report:')
#print(classification_report(y_test, y_pred))

# 9. Testing the Model with Custom Messages
# You can input your own message here to test if it's spam or not
sample_message = ['Congratulations! You have won a Rs.1000 Dmart coupon.']
sample_message_vectorized = vectorizer.transform(sample_message)
prediction = model.predict(sample_message_vectorized)
print(f'The message is {"spam" if prediction[0] == 1 else "ham"}')
```
