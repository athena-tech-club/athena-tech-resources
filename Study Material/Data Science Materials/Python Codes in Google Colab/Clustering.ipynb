```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

data = pd.read_csv('Iris.csv', header='infer').values
X = data[:, 1:-1]
y = data[:, -1]

test_split = float(input('Enter the test size: '))
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_split, stratify=y)

k = int(input('Enter the number of clusters you want to divide in: '))
n = int(input('Enter number of iterations you want to run: '))
centroids = np.zeros(shape=(k, X_train.shape[1]))
print(centroids)
per = np.random.permutation(X_train.shape[0])

for i in range(k):
    centroids[i, :] = X_train[per[i], :]

print(centroids)

for it in range(n):
    dist = np.zeros(shape=(k, X_train.shape[0]))
    for i in range(k):
        dist[i, :] = np.sqrt(np.sum((X_train - centroids[i]) ** 2, axis=1))
    membership = np.argmin(dist, axis=0)
    print(membership)
    for i in range(k):
        centroids[i, :] = np.mean(X_train[membership == i, :], axis=0)

print(centroids)
```
