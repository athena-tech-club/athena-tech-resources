```python
import numpy as np
from PIL import Image 
import matplotlib.pyplot as plt
from skimage import io
```

```python
import os
 
camera = io.imread('cameraman.tif')
print(camera.shape)#number of pixel
print(camera.ndim)# 2 dimentional
print(type(camera))#numpy arrray
print(camera.dtype)# integer value range from 0 to 255 (0 for black 255 for white)
import matplotlib.pyplot as plt
plt.imshow(camera,cmap=plt.cm.gray)
print(camera)
```

Images are stored as pixels values, each pixel value represents a color intensity value. 
Histograms are frequency distribution of these intensity values that occur in an image.

```python
ax = plt.hist(camera.ravel(), bins = 256)
plt.xlabel('Intensity Value')
plt.ylabel('Count')
plt.show()
```

A Gaussian filter is a linear filter. 
It's usually used to blur the image or to reduce noise.The Gaussian filter will blur edges and reduce contrast.


```python
import scipy
from scipy import ndimage
lena = io.imread('lena_gray.gif')
f = lena.astype(float)
blurred_f = ndimage.gaussian_filter(f, 3)
plt.figure(figsize=(12, 4))
plt.subplot(121)
plt.imshow(f, cmap=plt.cm.gray)
plt.axis('off')
plt.subplot(122)
plt.imshow(blurred_f, cmap=plt.cm.gray)
plt.axis('off')
plt.tight_layout()
plt.show()
```

The Sobel operator, sometimes called Sobel filter, is used in image processing and computer vision, particularly within edge detection algorithms where it creates an image emphasising edges. 

```python
from skimage.filters import sobel
im = io.imread('lena_gray.gif')
edge_sobel = sobel(im)
plt.figure(figsize=(12,4))
plt.subplot(121) 
plt.axis('off')
plt.imshow(im,cmap=plt.cm.gray )
plt.subplot(122)
plt.axis('off')
plt.imshow(edge_sobel, cmap=plt.cm.gray )
plt.tight_layout()
plt.show()
```

```python
#Convert RGB to Gray scale image 
from skimage.color import rgb2gray
original = io.imread('lena_color.gif')
grayscale = rgb2gray(original)
plt.figure(figsize=(12,4))
plt.subplot(121) 
plt.imshow(original,cmap=plt.cm.gray )
plt.subplot(122)
plt.imshow(grayscale, cmap=plt.cm.gray )
plt.tight_layout()
plt.show()
```

Extract and save frames from a video file

```python
# Importing all necessary libraries 
import cv2 
import os 
  
# Read the video from specified path 
cam = cv2.VideoCapture("traffic2.mp4") 

currentframe = 0
while(True): 
      
    # reading from frame 
    ret,frame = cam.read() 
  
    if ret: 
        # if video is still left continue creating images 
        name = 'frame' + str(currentframe) + '.jpg'
        print ('Creating...' + name) 
  
        # writing the extracted images 
        cv2.imwrite(name, frame) 
  
        # increasing counter so that it will 
        # show how many frames are created 
        currentframe += 1
    else: 
        break

# Release all space and windows once done 
#cam.release() 
#cv2.destroyAllWindows() 
```

```python
#Create video form the images store in data folder 
import cv2
import numpy as np
import glob
 
img_array = []
for filename in glob.glob('*.jpg'): # Read all the jpg file 
    img = cv2.imread(filename)
    height, width, layers = img.shape
    size = (width,height)
    img_array.append(img)
 
 
out = cv2.VideoWriter('project.avi',cv2.VideoWriter_fourcc(*'DIVX'), 30, size)
                       # Filename,      encoder Techniques,       No ofFrame
for i in range(len(img_array)):
    out.write(img_array[i])
out.release()
```
