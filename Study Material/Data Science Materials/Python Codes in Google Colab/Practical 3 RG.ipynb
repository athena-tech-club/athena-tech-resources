    df3.loc[:,"Embarked"] selects the "Embarked" column of the DataFrame df3.

So, if df3.loc[:,"Embarked"] contains values such as ["S", "C", "S", "Q"], then df3.loc[:,"Embarked"].unique() would return ["S", "C", "Q"], which is an array of the unique values in the "Embarked" column.


```python
df3.loc[df3.loc[:,"Age"].isna(),"Age"]=21
df3.info()
#selects the "Age" column of the DataFrame df3
#df3.loc[:,"Age"].isna() creates a boolean mask that is True for any missing values
#df3.loc[df3.loc[:,"Age"].isna(),"Age"] uses the boolean mask to select only the
#rows in the "Age" column that have missing values and the n sets to 21.
print(df3.loc[:,"Embarked"].unique())
df3.loc[df3["Embarked"].isna(),"Embarked"]='S' #replacing NA with "S"
df3.info()

df3.loc[:,"Cabin"].unique()
df3.loc[df3["Cabin"].isna(),"Cabin"]='C85' #replacing NA with "C85"
df3.info()
```

    df.loc[~df["Age"].isna(),"Age"] uses the inverted boolean mask to select only the rows in the "Age" column of the DataFrame df that have non-missing values.

```python
import numpy as np
df4=df.copy()
meanage=np.mean(df4.loc[~df4["Age"].isna(),"Age"].values) 
print(meanage)
df4.loc[df4["Age"].isna(),"Age"]=meanage
print(df4.info())

mfcabin=df.loc[:,"Cabin"].mode()[0]
#finds the mode (most frequent value) of the "Cabin" column of a DataFrame 
#called df and assigns it to the variable mfcabin.
print(mfcabin)
df4.loc[df["Cabin"].isna(),"Cabin"]=mfcabin
print(df4.info())

mfembarked=df.loc[:,"Embarked"].mode()[0]
print(mfembarked)
df4.loc[df["Embarked"].isna(),"Embarked"]=mfembarked
print(df4.info()) 
```

This code selects the values in the "Age" column of the DataFrame where the "Survived" column is equal to 0 (indicating that the passenger did not survive) and the "Age" column is not null (not missing). The tilde (~) before the "df['Age'].isna()" condition negates it, so the code selects only the rows where "Age" is not missing.

```python
meanage0=np.mean(df.loc[(~df["Age"].isna()) & (df["Survived"]==0), "Age"].values)
meanage1=np.mean(df.loc[(~df["Age"].isna()) & (df["Survived"]==1), "Age"].values)
print(meanage0)
print(meanage1)
df5.loc[df["Age"].isna() & df["Survived"]==0,"Age"]=meanage0
df5.loc[df["Age"].isna() & df["Survived"]==1,"Age"]=meanage1
```

select the "Cabin" column values of passengers who survived (i.e., "Survived" column value is 1), and then finds the mode of those values using the .mode() method. The resulting mode is then accessed using the [0] index

```python
mfcabin0=df.loc[df["Survived"]==0,"Cabin"].mode()[0]
mfcabin1=df.loc[df["Survived"]==1,"Cabin"].mode()[0]
print(mfcabin0)
print(mfcabin1)
df5.loc[df["Cabin"].isna() & df["Survived"]==0,"Cabin"]=mfcabin0
df5.loc[df["Cabin"].isna() & df["Survived"]==1,"Cabin"]=mfcabin1
print(df5.info())
```

 select the "Embarked" column values of passengers who did not survive (i.e., "Survived" column value is 0), and then finds the mode of those values using the .mode() method. The resulting mode is then accessed using the [0] index.

```python
mfembarked0=df.loc[df["Survived"]==0,"Embarked"].mode()[0]
mfembarked1=df.loc[df["Survived"]==1,"Embarked"].mode()[0]
print(mfembarked0)
print(mfembarked0)
df5.loc[df["Embarked"].isna() & df["Survived"]==0,"Embarked"]=mfembarked0
df5.loc[df["Embarked"].isna() & df["Survived"]==1,"Embarked"]=mfembarked1
print(df5.info())
```
