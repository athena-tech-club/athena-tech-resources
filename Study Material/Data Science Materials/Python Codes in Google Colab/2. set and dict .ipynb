# **sets in python ** A set is a collection which is unordered, unchangeable*, and unindexed.

```python
myset = {"apple", "banana", "cherry"}
print(myset)
```

```python
#duplicates check
thisset = {"apple", "banana", "cherry", "apple"}
print(thisset)
```

```python
#set constructor
# set() in round
thisset = set(("apple", "banana", "cherry")) # note the double round-brackets
print(thisset)
```

```python
#access Set items
thisset = {"apple", "banana", "cherry"}
for x in thisset:
  print(x)
```

```python
#check for items in set
thisset = {"apple", "banana", "cherry"}
print("pineapple" in thisset)
```

```python
#add set items
thisset = {"apple", "banana", "cherry"}
thisset.add("orange")
print(thisset)
```

```python
#use update method
thisset = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}
thisset.update(tropical)
print(thisset)
```

```python
#use loop Set
thisset = {"apple", "banana", "cherry"}
for x in thisset:
  print(x)
```

```python
#join two sets
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2)
print(set3)
```

```python
#update the set
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set1.update(set2)
print(set1)
```

```python
#use intersection update
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}
x.intersection_update(y)
set3=x.intersection(y)
print(set3)
print(x)
```
