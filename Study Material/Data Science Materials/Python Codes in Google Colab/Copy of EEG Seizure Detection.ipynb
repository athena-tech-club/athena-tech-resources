```python
# 1. Logistic Regression

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

data = pd.read_csv('data.csv')
X = data.iloc[:, 1:-1]
y = data.iloc[:, -1]

model = LogisticRegression()
model.fit(X, y)

coefficients = model.coef_[0]

importance_scores = np.abs(coefficients)

plt.figure(figsize=(12, 6))
plt.bar(range(len(X.columns)), importance_scores)
plt.xlabel('Feature')
plt.ylabel('Absolute Coefficient Value (Feature Importance)')
plt.title('Feature Importance for Logistic Regression')
plt.xticks(range(len(X.columns)), X.columns, rotation=90)
plt.tight_layout()
plt.show()

feature_logistic1 = sorted(list(zip(X.columns, importance_scores)), key=lambda x: x[1], reverse=True)[:30]
feature_logistic = list(zip(X.columns, importance_scores))
feature_logistic
```

```python
# 2. Random Forest

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier

data = pd.read_csv('data.csv')

X = data.iloc[:, 1:-1]
y = data.iloc[:, -1]

clf = RandomForestClassifier(n_estimators=100, random_state=0)
clf.fit(X, y)

importances = clf.feature_importances_

sorted_indices = np.argsort(importances)[::-1]
sorted_features = X.columns[sorted_indices]
sorted_importances = importances[sorted_indices]

plt.figure(figsize=(12, 6))
plt.bar(range(len(sorted_features)), sorted_importances)
plt.xlabel('Feature')
plt.ylabel('Importance')
plt.title('Feature Importance for Random Forest')
plt.xticks(range(len(sorted_features)), sorted_features, rotation=90)
plt.tight_layout()
plt.show()

feature_rf1 = list(zip(sorted_features, sorted_importances))
feature_rf = list(zip(X.columns, importances))
feature_rf
```

```python
# 3. Decision Tree Classifier

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

data = pd.read_csv('data.csv')

X = data.iloc[:, 1:-1]
y = data.iloc[:, -1]

clf = DecisionTreeClassifier(random_state=0)
clf.fit(X, y)

feature_importances = clf.feature_importances_

plt.figure(figsize=(12, 6))
plt.bar(range(len(feature_importances)), feature_importances)
plt.xlabel('Feature')
plt.ylabel('Feature Importance')
plt.title('Feature Importance for Decision Tree Classifier')
plt.xticks(range(len(feature_importances)), X.columns, rotation=90)
plt.tight_layout()
plt.show()

feature_tree1 = sorted(list(zip(X.columns, feature_importances)), key=lambda x: x[1], reverse=True)
feature_tree = list(zip(X.columns, feature_importances))
feature_tree
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

data = pd.read_csv('data.csv')

X = data.iloc[:, 1:-1]
y = data.iloc[:, -1]

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

feature_importance = model.coef_

plt.figure(figsize=(12, 6))
plt.bar(range(len(feature_importance)), feature_importance)
plt.xlabel('Feature')
plt.ylabel('Feature Importance')
plt.title('Feature Importance for Decision Tree Classifier')
plt.xticks(range(len(feature_importance)), X.columns, rotation=90)
plt.tight_layout()
plt.show()

feature_poly1 = sorted(list(zip(X.columns, feature_importance)), key=lambda x: x[1], reverse=True)
feature_poly = list(zip(X.columns, feature_importance))
feature_poly
```

```python
# 5. XGBOOST

import pandas as pd
import xgboost as xgb
import matplotlib.pyplot as plt

data = pd.read_csv('data.csv')

X = data.iloc[:, 1:-1]
y = data.iloc[:, -1]

model = xgb.XGBRegressor(objective='reg:squarederror')
model.fit(X, y)

importance = model.feature_importances_

feature_names = X.columns

plt.figure(figsize=(12, 6))
plt.bar(feature_names, importance)
plt.xlabel('Feature Importance')
plt.ylabel('Features')
plt.title('XGBoost Feature Importance')
plt.tight_layout()
plt.show()

feature_XGBoost1 = [(feature, importance_score) for feature, importance_score in zip(feature_names, importance)]
feature_XGBoost1.sort(key=lambda x: x[1], reverse=True)

feature_XGBoost = [(feature, importance_score) for feature, importance_score in zip(feature_names, importance)]
feature_XGBoost
```

```python
# 6. SIMPLE SVM

import pandas as pd
import numpy as np
from sklearn import svm
import matplotlib.pyplot as plt

data = pd.read_csv('data.csv')

X = data.iloc[:, 1:-1]
y = data.iloc[:, -1]

model = svm.SVC(kernel='linear')
model.fit(X, y)

coefficients = np.abs(model.coef_.flatten())

sorted_indices = np.argsort(coefficients)[::-1]
feature_names = X.columns
sorted_feature_names = [feature_names[i] for i in sorted_indices]
sorted_coefficients = [coefficients[i] for i in sorted_indices]

plt.figure(figsize=(12, 6))
plt.bar(range(len(sorted_feature_names)), sorted_coefficients, tick_label=sorted_feature_names)
plt.xlabel('Feature')
plt.ylabel('Absolute Coefficient Magnitude')
plt.title('Feature Importance for SVM')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

feature_svm = list(zip(sorted_feature_names, sorted_coefficients))
feature_svm
```
